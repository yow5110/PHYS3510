# 2. Types

There are three different kinds of numerical objects in Python - a more formal definition of objects later in the course. Most of the examples above deal with integer numbers. To check the type of a quantity in python you can use the built-in type() function. For example, we have **integters**:

```python
type(1)
```

Not all real numbers can be represented on a computer, due to the fact that a fixed amound of memory is used to store each number (i.e. we can only save a fixed number of digits). Floating point notation is the common way to store numbers in a computer program, for this reason Python calls its representation of real numbers **float**. 

```python
type(3.1)
```

```python
type(3e3)
```

(Side note. Floating point notation is closely related to the scientific notation we are familiar with, where we use a significand and an exponent to represent a real number. For example in base 10, 1234.56 = 123456 × 10$^{-2}$. So in a computer this would be stored as 123456 along with -2. The name "floating" refers to the fact that the decimal point could float to any position by changing the exponent without changing the integer significand.)

The last numerical type that we can use in python corresponds to **complex numbers**. The imaginary unit is written with the engineering notation, using the letter j. Note that we don't have to specify the multiplication between the imaginary part of the complex number and the imaginary unit.

```python
type(3+1j)
```

We can also check that things inside quotations are **strings** in Python

```python
type("hello")
```
