There are more than one way of importing functions from modules. Here are some more variations - at the end we will recommend one as the usual convention. 

You can import one single function from a module using "from [module name] import [function name]. This will not import other functions in this module.

If you don't want to keep writing "numpy." in principle you could import all the functions of a module using an asterisk *

However, this is NOT recommended. With the above command you may be importing a large number of functions that could potentially conflict with functions with the same name in the built-ins or in another module. To avoid confusion, it is always a good idea to keep track of the modules as they are imported. 

To achieve this while not having to type long module names, the recommended convention is "import [module name] as [short alias for module]" and then calling functions using "[short module alias].[function name]()". You can choose any short alias you want. For numpy, a common choice is

So the important trick to remember here is that **we should never check if two floats are equal using a == operation!**

In decimal form (which we are familiar with), when we express fractions like 1/2 and 1/4, we get nice decimal forms like 0.5 and 0.25. But for 1/3 = 0.3333... we have to truncate it somewhere, creating an approximation like 1/3 $\approx$ 0.33333333333.

By the same token, in binary form, some fractions become easier to express and some needs approximating. For example, 1/10 (in decimal) converted to binary becomes 0.0 0011 0011 0011 0011 .... So in  base 2 we have to truncate it somewhere, creating an approximation. You can check for your self that

```python
2**-4 + 2**-5 + 2**-8 + 2**-9 + 2**-12 + 2**-13 + 2**-16 + 2**-17 + 2**-19
```

The same applies to, for example, 0.2. So that if you want to do 0.1+0.2, even before the calculation, the numbers are already approximated. The end result, which you know should be 0.3, will be output from python as the sum of these two approximate numbers, and happens to be slightly larger than 0.3.
