# Mutable and Immutable
Most of the things that we have seen in Python so far are **immutable** objects: integers, floats, complex numbers, Booleans, and strings – meaning that you cannot change their values. If you think about it, it makes sense in that the number 5 is always the number 5. If you change it it will be a different number, so there's nothing to be changed about the number 5 itself. It is in this sense that we call all integers, floats, Booleans, complex numbers, and strings in Python **immutable**.

But wait! We've defintely seen cases where we assigned $x$=5 and we're obviously allowed to update $x$ to another number. How is the number $x$ is immutable? To explain this we need to go a bit deeper. 

When an immutable object like $x$=5 is created in python, $x$ is stored in memory at a fixed location. The assignment operator tells python that the location of the variable $x$ is the same as the location of the number 5. We can see this behavior by exploiting the built-in **id()** function, which shows the unique identifier of a python object within the current program.  

```python
x=5
```

```python
id(5)
```

```python
id(x)
```

As expected the ID of the integer number 5 and of the variable $x$ are the same. What may seem less trivial is that, if we execute a second assignment statement for a different variable y, we still get the same ID. 

```python
y=5
id(y)
```

This is beacuse the integer 5 was already present in our program and had already a unique identifier. We merely created a different label $y$ for the same quantity. 

If we perform an operation on y and save the result in y, we see that the identifier of y changed. 

```python
y=y+1
print(y,id(y))
```

This is beacuse the operation on the right of the assignment produces the integer value of 6. Since this value was not in the program before, now python needs to point the location of $y$ to the location of the number 6, which is different from its starting location. 

Python behaves in this way with any instance of an immutable class. Int, float, complex, bool are all immutable classes. Also strings are immutable in python, so the same behavior applies to them. 

```python
c='hello'
d='hello'
print(id(c),id(d))
```

By contrast, Python **lists** are **mutable** objects. This feature allows lists to change their size (i.e. the number of components) and their content. The additional flexibility requires python to deal with lists in a different way.

```python
a=[0, 2, 4]
```

```python
id(a)
```

If we modify the list, the identifier does not change

```python
a.append(5)
print(a)
print(id(a))
```

This implies that if we pass the list $a$ onto a new list $b$, we are only passing along their ID

```python
b=a
print(b)
print(id(b))
```

so that lists $a$ and $b$ are now the exact same thing, and any action performed on $b$ will also update $a$

```python
b[0]=3
print(b)
print(a)
print(id(b),id(a))
```

Assignemnts of immutable objects really act as aliases (you are just using a different name to access the same actual object). If we want to perform actions on a list without affecting the original list, we need to create a separate copy of the object. **This can be achieved with the copy() method**. 

```python
b=a.copy()
print(id(a),id(b))
```

```python
b[0]=7
print(a)
print(b)
```

So the main takeaway here is: **use the copy() method when copying lists**
