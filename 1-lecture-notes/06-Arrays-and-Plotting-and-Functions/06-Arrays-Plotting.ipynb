we'll continue working on our first plot. Through this practice we'll introduce **list comprehension**, **new plotting tools**, **custom functions**, and **numpy arrays**

```python
import numpy as np
epsilon=4.0e-4
sigma=6.0
rmin = 5.5
rmax = 10
dr = 0.001
r_grid = np.arange(rmin, rmax, dr)

V_list = [ ] #initiate an empty list
zero_list = []      
constant_list = []  

for r in r_grid:    
    V =  4.0 * epsilon * ((sigma/r)**12-(sigma/r)**6)
    V_list.append( V )
    zero_list.append(0)             #<<<<<
    constant_list.append(-epsilon)  #<<<<<
    
import matplotlib.pyplot as plt
plt.plot( r_grid , V_list)
plt.plot( r_grid , zero_list)
plt.plot( r_grid , constant_list)

plt.show()
```
