## Conditional Blocks (if statement)
One of the most important components of any algorithm or computer program is the possibility to take different actions  according to a specific state of the system (branching). Python uses the if statement to provide this possibility. There are many ways of writing an if statement. The easiest way only uses an "if [condition]:" statement followed by a set of statement that are executed only if the condition evaluates to True. 

Note that **all the indented statements**  following the if statement belongs to the if block. All these statements are executed if and only if the condition in the if statement evaluates to True. 

We can trying messing with the if block and see that anything following the if statement without indentation does not belong to the if block, and therefore is always executed, even when the condition in the if block evaluates to False.

If we have only one statement, we can use a shorter notation reporting the statement in the same line as the if. 

In many situations, we may want to execute alternative statments if the condition in the conditionals are not satisfied. The syntax in python require us to specify an else : statement.

Also in this case, we can have as many statements as we like after the else block, but they need to be identified as a block of code using indentation. 

We can be even more flexible in our branching construct by specifying additional alternative conditions for branching using the **elif** statement. We can insert as many elif blocks as we need.
