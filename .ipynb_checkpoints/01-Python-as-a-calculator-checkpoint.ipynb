Simple physics problem example. Energy of photon = 1.9 eV.  What's its wavelength? 

$E = hc/\lambda$
so $\lambda = h c / E $.

Including appropriate conversion factors (from eV to J and from m to nm), we have 

```python
6.63*10**-34 * 3*10**8 / (1.9 * 1.6*10**-19) * 10**9
```

When in doubt, use parentheses

```python
(6.63*10**-34) * (3*10**8) / (1.9 * 1.6*10**-19) * 10**9
```

What if we carry out more sophisticated calculations? Neater way - use variables.

```python
h = (6.63*10**-34)
c = 3 * 10**8
J2Ev = 1.6*10**-19
Energy = 1.9
m2nm = 10**9

h * c / (Energy * J2Ev) * m2nm
```

If you copy all these commands into a text file - then that's a python script!

Once you start writing scripts, make sure to leave comments so that your code is readable.

```python
# defining variable useful for the calculation
h = 6.626e-34
c = 3e8
energy = 1.9
eV2J = 1.6e-19  
m2nm = 1e-9

# convert photon energy in eV into wavelength in nm
L = h * c / (energy * eV2J) / m2nm
print("The wavelength of a photon with energy ", energy, "eV is ",L, " nm")
```

## More on the equal sign

It may seems harmless to write equations like $a=b$ as we are familiar with in maths and physics. However in Python, you'll see the following code produce an error of "NameError: name 'b' is not defined"

```python
#try this
a = b
```

In Python (and in almost all programming languages), "=" really means assign. We assign the value on the right to the variable on the left.

```python
#another to show this
a = 4.2
print(a, " before")

a = a + 1.0
print(a, " after")
```

Comment: this shows that the computer has an internal **state** that keeps track of all variables.

In Spyder you can check the state of all variables in the **variables  panel**.

Sometimes we $do$ want to use the equal sign to test if two values are the same. In these cases we use "=="

```python
a = 2  #assign a to be 2.
a == 2  #check if a is equal to 2.
```

We can even assign the result of evaluating whether $a$ is equal to 2 to another variable!

```python
a=2
b=3

Check_a  = (a==2)
Check_b =  (b==2)
print(Check_a)
print(Check_b)
```
